```python
# Importing libraries 

import pandas as pd  
import matplotlib.pyplot as plt
import seaborn as sns

from textblob import TextBlob #to understand the meaning of the sentences
from wordcloud import WordCloud

```

```python
# Load the dataset

data = pd.read_csv("Sentiment_Analysis .csv")

data.head()

```

```python
# Define the sentiment analysis functions

def get_sentiment(text):
    meaning = TextBlob(text)

    if meaning.sentiment.polarity >0:
        return "Positive"

    elif  meaning.sentiment.polarity <0:
        return "Negative"

    else:
        return "Neutral"    
        
```

```python
data["Sentiment"] = data["Sentences"].apply(get_sentiment)
```

```python
data.head(8)
```

```python
data.to_excel("Sentiment_Analysis_Mratyunjai.xlsx",index= False)
```

```python
#Filter only negetive reviews 

negative_reviews = data[data["Sentiment"] == "Negative"]
```

```python
negative_reviews.head(10)
```

```python
# Combine all negative sentences into one srting/ paragraph

negative_para = " ".join(negative_reviews["Sentences"])

print(negative_para)
```

```python
# Create the Word Cloud for the negative para

wordcloud = WordCloud(width=800, height=400, background_color="Black").generate(negative_para)

#plot the Word Cloud

plt.figure(figsize=(10,6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.title("Most Common Words in Negative Sentences")
plt.axis("off")
plt.show()


```
